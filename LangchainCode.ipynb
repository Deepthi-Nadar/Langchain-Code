```python
from langchain_groq import ChatGroq

llm = ChatGroq(api_key=api_key, model=llama_model)
```

```python
def simple_chat_without_memory(user_query):
  response=llm.invoke(user_query)
  return response.content
```

##Store the Human and AI conversation


```python
from langchain_core.messages import HumanMessage, AIMessage
```

```python
convo=[]

def simple_chat_with_memory(user_query):
  convo.append(HumanMessage(content=user_query))
  response=llm.invoke(convo)
  convo.append(AIMessage(content=response.content))

  return response.content
```

```python
print(simple_chat_with_memory("i want to learn ai"))
```

```python
print(convo)
```

```python
print(simple_chat_with_memory("can you give me a learning plan "))
```

# User session and prompt for llm

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate(messages=[
    ("system", "You are a nice AI bot that helps a user figure out what to eat in one short sentence"),
    ("human", "{user_input}") ])

chain = prompt | llm

response=chain.invoke({"user_input": "I like noodles"})
print(response)
```

User session and conversion

```python
prompt_w_memory = ChatPromptTemplate.from_messages([
    ("system", "You are a nice AI bot that helps a user figure out what to eat in one short sentence"),
    ("placeholder", "{history}"),
    ("human", "{user_input}")
])

chain_w_memory = prompt_w_memory | llm
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
```

```python
store = {}

def get_history(session_id: str):
  if session_id not in store:
    store[session_id] = ChatMessageHistory()
  return store[session_id]
```

```python
for i in store:
  print(i)
```

```python
from langchain_core.runnables import RunnableWithMessageHistory
```

```python
chat_w_memory = RunnableWithMessageHistory(
    runnable=chain_w_memory,
    get_session_history=get_history,
    input_message_key="user_input",
    history_message_key="history"
)

response=chat_w_memory.invoke({"user_query":"I like noodles"}, {"configurable":{"session_id":"user1"}})
print(response)
```

```python
response=chat_w_memory.invoke({"user_query":"which one would be the best"}, {"configurable":{"session_id":"user1"}})
print(response)
```
